## Section: Nuclear magnetic resonance

```python
# gyromagnetic ratios
# proton gyromagnetic ratio
# 2.675 221 8708(11) x 10^8 s-1 T-1
displayed_gamma_p = (gamma_p / (2 * PI)).to(unit.MHz / unit.T)
displayed_gamma_p_uncertainty = (gamma_p.uncertainty * gamma_p.unit / (2 * PI)).to(unit.MHz / unit.T)
print(
    f"Proton gyromagnetic ratio: {displayed_gamma_p:.11g} +/- {displayed_gamma_p_uncertainty:.9f}"
)
```

```python
# at high-temperature limit $\Delta E \ll k_B T$ 
# the fractional spin polarization is
B_0 = 31.67 * unit.millitesla
temp = 180 * unit.K
pol = (const.hbar * gamma_p / unit.rad * B_0 / (2 * const.k_B * temp)).to(unit.one)
print(f"proton thermal polarization = {pol:.0e}")
```

```python
# at high-temperature limit $\Delta E \ll k_B T$
# the fractional spin polarization is
B_0 = 1 * unit.tesla
temp = 300 * unit.K
pol = (const.hbar * gamma_p / unit.rad * B_0 / (2 * const.k_B * temp)).to(unit.one)
print(f"proton thermal polarization = {pol:.1e}")
```

```python
# spin projection noise (SPN)
sample_volume = 1 * unit.centimeter**3
sample_molar_mass = 18 * unit.gram / unit.mol  # water
sample_density = 1 * unit.gram / unit.centimeter**3  # water
n_protons = (sample_volume * sample_density / sample_molar_mass * 2 * const.N_A).to(unit.one)  # number of protons in 1 cm^3 of water

# polarization
B_0 = 1 * unit.tesla
temp = 300 * unit.K
pol = (const.hbar * gamma_p / unit.rad * B_0 / (2 * const.k_B * temp)).to(unit.one)
# print(f"proton thermal polarization = {pol:.1e}")

M_eqb = (pol * mu_p * (pol * n_protons) / sample_volume).to(unit.ampere / unit.meter)
M_SPN = (mu_p * np.sqrt(n_protons) / sample_volume).to(unit.ampere / unit.meter)
print(f"number of protons in 1 cm^3 of water = {n_protons:.1e}")
print(f"thermal-equilibrium magnetization = {M_eqb:.1e}")
print(f"spin projection noise (SPN) in magnetization units = {M_SPN:.1e}")
```

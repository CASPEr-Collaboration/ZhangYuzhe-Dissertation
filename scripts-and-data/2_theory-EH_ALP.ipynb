#### ALP field gradient

##### Maximum surface speed of the Earth
##### and 
##### Estimation of the Lorentz-boost correction to the ALP field gradient


```python
T_earth = 8.62e4 * unit.s
v_rot_max = 2 * np.pi * unit.earthRad / T_earth
print(
    f"Maximum surface speed of the Earth: { v_rot_max.si:.0f} = {v_rot_max.to(const.c):.2e}",
)

gamma_lab = 1 / (1 - (v_rot_max / const.c) ** 2) ** 0.5
print(f"Earth's surface Lorentz factor: {gamma_lab:.2e}")

# first correction term in Equation~\eqref{eq:gradient_boost_terms_1}
print(
    f"First correction term in Equation~\\eqref{{eq:gradient_boost_terms_1}} approximately = {gamma_lab - 1:.2e}"
)

# second correction term in Equation~\eqref{eq:gradient_boost_terms_2}
nu_a = 1 * unit.MHz
correction_term_2_nabla_a = gamma_lab * (
    2 * np.pi * nu_a * v_rot_max / const.c**2 * unit.earthRad
)
print(
    f"Second correction term in Equation~\\eqref{{eq:gradient_boost_terms_2}} approximately = {correction_term_2_nabla_a.si:.2e} |nabla a|"
)
```

##### Estimation of the ALP field gradient

```python
rhoE_DM = 1e16 * 0.3 * unit.GeV / unit.cm**3
DM_mass_by_moon_orbit = 4e-9 * unit.earthMass
m_a = const.h * 1.348 * unit.MHz / const.c**2
N_a = DM_mass_by_moon_orbit / m_a
g_aNN = 1e-9 * unit.GeV ** (-1)
# {eq:a0_rhoDM_Psi_real}
a_0_rho = (2 * const.hbar**3 * rhoE_DM / (m_a**2 * const.c)) ** 0.5
# {eq:a0_Psi_real}
a_0_Na = (N_a * const.hbar**3 * const.c / (2 * m_a)) ** 0.5 * (
    1 * unit.earthRad ** (-1.5)
)
theta_gradient = 1 * unit.earthRad ** (-2.5)
# {eq:gradient_Psi_sum}
Omega_a_by_gradient = (
    const.c
    * g_aNN
    * (N_a * const.hbar**3 * const.c / (2 * m_a)) ** 0.5
    * theta_gradient
)
v_esc = 11.2 * unit.km / unit.s
Omega_a_by_speed = (0.5/const.hbar) * const.c * g_aNN * a_0_rho * m_a * v_esc

print("number of ALPs:", N_a.to(unit.one))
print("ALP field amplitude a_0 (estimated by rho_DM):", a_0_rho.si)
print("ALP field amplitude a_0 (estimated by N_a):", a_0_Na.si)
print("ALP wavefunction theta gradient:", theta_gradient)
print(f"Rabi frequency (estimated by gradient): 2 pi * {Omega_a_by_gradient.to(unit.mHz)/(2 * np.pi):.1f}")
print(f"Rabi frequency (estimated by speed): 2 pi * {Omega_a_by_speed.to(unit.mHz)/(2 * np.pi):.1f}")
```

```python
# estimation of the dB wavelength
scan_freq_max = 600 * unit.MHz
scan_mass_600MHz = const.h * scan_freq_max / const.c**2
scan_mass_4MHz = const.h * 4 * unit.MHz / const.c**2
scan_mass_1MHz = const.h * 1 * unit.MHz / const.c**2
scan_mass_1kHz = const.h * 1 * unit.kHz / const.c**2
v_esc = (2 * np.abs(earth_pot_unit_mass)) ** 0.5

lambda_dB_600MHz = (const.h / (scan_mass_600MHz * v_esc)).si
lambda_dB_4MHz = (const.h / (scan_mass_4MHz * v_esc)).si
lambda_dB_1MHz = (const.h / (scan_mass_1MHz * v_esc)).si
lambda_dB_1kHz = (const.h / (scan_mass_1kHz * v_esc)).si

print("Escape velocity at Earth surface:", v_esc.to(unit.km / unit.s))

# 600 MHz
print("ALP mass at 600 MHz Compton freq:", f"{(scan_mass_600MHz * const.c**2).to(unit.eV):.1e}")
print("de Broglie wavelength at 600 MHz:", f"{lambda_dB_600MHz.to(unit.km):.1e}")

# 4 MHz
print("ALP mass at 4 MHz Compton freq:", f"{(scan_mass_4MHz * const.c**2).to(unit.eV):.1e}")
print(
    "de Broglie wavelength at 4 MHz:",
    f"{lambda_dB_4MHz.to(unit.km):.1e}",
    "=",
    f"{lambda_dB_4MHz.to(unit.earthRad):.1e}",
)

# 1 MHz
print("ALP mass at 1 MHz Compton freq:", f"{(scan_mass_1MHz * const.c**2).to(unit.eV):.1e}")
print(
    f"de Broglie wavelength at 1 MHz: {lambda_dB_1MHz.to(unit.m):.1e}",
    "=",
    f"{lambda_dB_1MHz.to(unit.earthRad):.1e}",
)

# 1 kHz
print("ALP mass at 1 kHz Compton freq:", f"{(scan_mass_1kHz * const.c**2).to(unit.eV):.1e}")
print(
    f"de Broglie wavelength at 1 kHz: {lambda_dB_1kHz.to(unit.m):.1e}",
    "=",
    f"{lambda_dB_1kHz.to(unit.earthRad):.1e}",
)
```

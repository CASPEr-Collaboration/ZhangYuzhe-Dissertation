## Section: Sample and sample insert

##### Thermal polarization and magnetization of proton spins

```python
# Sample: liquid methanol
name = "C-12 Methanol"
gamma = gamma_p
massDensity = 0.792 * unit.g * unit.cm ** (-3)
molarMass = 32.04 * unit.g / unit.mol
numOfSpinsPerMolecule = 4

mu = mu_p
temp = 180 * unit.K
B_0 = 31.67 * unit.mT
I_p = 1 / 2
# proton spin number density
numDensity = (massDensity / molarMass * numOfSpinsPerMolecule * const.N_A).to(
    unit.cm ** (-3)
)
print(f"Proton number density: {numDensity:.2e}")
M_0 = (
    numDensity
    * gamma**2
    * const.hbar**2
    * I_p
    * (I_p + 1)
    * B_0
    / (3 * const.k_B * temp)
).to(unit.A * unit.m ** (-1), unit.dimensionless_angles())
print(f"Thermal polarization: {(M_0 / (numDensity * mu)).to(unit.one):.2e}")
print(f"Magnetization: {M_0:.2e}")
```

## Section: Pickup coil

##### filling factor

```python
height_central_coil = 24 * unit.mm
radius_central_coil = 14 * unit.mm
volume_coil = (np.pi * radius_central_coil**2 * height_central_coil).to(unit.cm**3)

height_sample = 24 * unit.mm
radius_sample = 4 * unit.mm
volume_sample = (np.pi * radius_sample**2 * height_sample).to(unit.cm**3)
filling_factor = (volume_sample / volume_coil).to(unit.percent)

print(f"Volume of central coil: {volume_coil:.2e}")
print(f"Volume of sample: {volume_sample:.2e}")
print(f"Filling factor: {filling_factor:.1f}")
```

### 20221214 measurement

#### Scan range

```python
center_freq = 1.348585 * unit.MHz
scan_half_width = 12 * unit.Hz
scan_start = center_freq - scan_half_width
scan_stop = center_freq + scan_half_width

center_ALP_mass = (const.h * center_freq / const.c**2).to(unit.neV / const.c**2)
scan_half_width_ALP_mass = (const.h * scan_half_width / const.c**2).to(unit.neV / const.c**2)
print(f"center ALP mass: {center_ALP_mass:.6f}")
print(f"scan half width ALP mass: {scan_half_width_ALP_mass:.6f}")
```

#### Bias magnetic field

```python
center_B0 = center_freq / (gamma_p / 2 / PI)
center_B0 = center_B0.to(unit.mT)
scan_start_B0 = (scan_start / (gamma_p / 2 / PI)).to(unit.mT)
scan_stop_B0 = (scan_stop / (gamma_p / 2 / PI)).to(unit.mT)
print(f"center B0: {center_B0:.6f}")
print(f"scan start B0: {scan_start_B0:.6f}")
print(f"scan stop B0: {scan_stop_B0:.6f}")
```

#### Earth-bound ALP eigenstate table

The table `tab:EH_ALP-eigenstates` lists selected Earth-bound ALP eigenstates from low to high energy. The energy values below are the eigenenergy offsets relative to the ALP rest energy, so the frequency offset is calculated as

$\Delta\nu_a = \frac{E_\mathrm{eigen}-m_a c^2}{h}.$

The Rabi-frequency column is left as a placeholder until the final signal normalization and state-population assumptions are fixed.

```python
# Raw eigenstate information copied from tex/results.tex.
# Columns in the original note:
# n_r, l, principal n, name, eigenenergy offset E_eigen - m_a c^2 (eV),
# kinetic energy T (eV), mean velocity (m/s).
earth_bound_states = [
    {"n_r": 0, "l": 0, "principal_n": 1, "name": "1s", "eigen_E_eV": 8.194e-19, "kinetic_T_eV": 1.734e-20, "mean_v_m_s": 8.680e2},
    {"n_r": 1, "l": 0, "principal_n": 2, "name": "2s", "eigen_E_eV": 2.228e-18, "kinetic_T_eV": 7.611e-19, "mean_v_m_s": 5.751e3},
    {"n_r": 2, "l": 0, "principal_n": 3, "name": "3s", "eigen_E_eV": 3.130e-18, "kinetic_T_eV": 6.378e-19, "mean_v_m_s": 5.265e3},
    {"n_r": 0, "l": 1, "principal_n": 2, "name": "2p", "eigen_E_eV": 3.258e-18, "kinetic_T_eV": 3.983e-19, "mean_v_m_s": 4.160e3},
    {"n_r": 3, "l": 0, "principal_n": 4, "name": "4s", "eigen_E_eV": 3.702e-18, "kinetic_T_eV": 8.109e-19, "mean_v_m_s": 5.937e3},
    {"n_r": 4, "l": 0, "principal_n": 5, "name": "5s", "eigen_E_eV": 4.158e-18, "kinetic_T_eV": 7.940e-19, "mean_v_m_s": 5.875e3},
    {"n_r": 1, "l": 1, "principal_n": 3, "name": "3p", "eigen_E_eV": 4.222e-18, "kinetic_T_eV": 6.542e-19, "mean_v_m_s": 5.332e3},
    {"n_r": 2, "l": 1, "principal_n": 4, "name": "4p", "eigen_E_eV": 4.860e-18, "kinetic_T_eV": 5.147e-19, "mean_v_m_s": 4.730e3},
    {"n_r": 3, "l": 1, "principal_n": 5, "name": "5p", "eigen_E_eV": 5.158e-18, "kinetic_T_eV": 4.251e-19, "mean_v_m_s": 4.299e3},
    {"n_r": 4, "l": 1, "principal_n": 6, "name": "6p", "eigen_E_eV": 5.329e-18, "kinetic_T_eV": 3.285e-19, "mean_v_m_s": 3.779e3},
]

# Convert energy offsets to frequency offsets. Astropy's eV/Hz conversion
# requires the spectral equivalency because E = h nu.
for state in earth_bound_states:
    delta_nu = (state["eigen_E_eV"] * unit.eV).to(unit.mHz, equivalencies=unit.spectral())
    state["delta_nu_mHz"] = delta_nu.value

# Placeholder values for the Rabi-frequency column in tab:EH_ALP-eigenstates.
# Fill these after choosing the state-population model and normalization.
rabi_frequency_rms_mHz = {state["name"]: None for state in earth_bound_states}

earth_bound_states_sorted = sorted(earth_bound_states, key=lambda row: row["eigen_E_eV"])
for state in earth_bound_states_sorted:
    print(
        f"{state['name']:>2s}: "
        f"Delta nu_a = {state['delta_nu_mHz']:.3f} mHz, "
        f"Omega_a/(2pi) = {rabi_frequency_rms_mHz[state['name']]}"
    )
```

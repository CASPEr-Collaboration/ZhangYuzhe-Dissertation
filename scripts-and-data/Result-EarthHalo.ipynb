### 20221214 measurement

#### Scan range

```python
center_freq = 1.348585 * unit.MHz
scan_half_width = 12 * unit.Hz
scan_start = center_freq - scan_half_width
scan_stop = center_freq + scan_half_width

center_ALP_mass = (const.h * center_freq / const.c**2).to(unit.neV / const.c**2)
scan_half_width_ALP_mass = (const.h * scan_half_width / const.c**2).to(unit.neV / const.c**2)
print(f"center ALP mass: {center_ALP_mass:.6f}")
print(f"scan half width ALP mass: {scan_half_width_ALP_mass:.6f}")
```

#### Bias magnetic field

```python
center_B0 = center_freq / (gamma_p / 2 / PI)
center_B0 = center_B0.to(unit.mT)
scan_start_B0 = (scan_start / (gamma_p / 2 / PI)).to(unit.mT)
scan_stop_B0 = (scan_stop / (gamma_p / 2 / PI)).to(unit.mT)
print(f"center B0: {center_B0:.6f}")
print(f"scan start B0: {scan_start_B0:.6f}")
print(f"scan stop B0: {scan_stop_B0:.6f}")
```

#### Earth-bound ALP eigenstate table

The table `tab:EH_ALP-eigenstates` lists selected Earth-bound ALP eigenstates from low to high energy. The energy values below are the eigenenergy offsets relative to the ALP rest energy, so the frequency offset is calculated as

$\Delta\nu_a = \frac{E_\mathrm{eigen}-m_a c^2}{h}.$

The Rabi-frequency column is left as a placeholder until the final signal normalization and state-population assumptions are fixed.

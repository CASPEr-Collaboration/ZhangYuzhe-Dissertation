```python
print("earth mass", (1 * unit.earthMass).to(unit.kg))
print("earth radius", (1.0 * unit.earthRad).to(unit.km))
```

```python
# m_a_ueV = 1 * unit.ueV / const.c**2
earth_pot_unit_mass = -const.G * unit.earthMass / unit.earthRad
earth_pot_10eVtoc2 = np.abs(earth_pot_unit_mass * 10 * unit.eV / const.c**2).to(unit.eV)
# check(m_a_ueV.to(unit.kg))

print(
    "Gravitational potential per unit mass at Earth surface:",
    earth_pot_unit_mass.to(unit.MJ / unit.kg),
)
print("Gravitational potential for 10 eV/c^2 ALP at Earth surface:", earth_pot_10eVtoc2)
```

```python
# Units
# a(r, t)
# {eq:a0_Psi_real}
rhoE_DM = 0.3 * unit.GeV / unit.cm**3
m_a = 10 * unit.eV / const.c**2
# N_a = 1e6
a_0 = (2 * const.hbar**3 * rhoE_DM / (m_a**2 * const.c))**0.5 
print("a_0 for 10 eV/c^2 ALP:", (a_0).si)
print("ALP field gradient unit:", (a_0 / unit.m**2.5).to(unit.eV / unit.m**2.5))
print("Rabi frequency unit:", (const.hbar * const.c * 1e-9 * unit.GeV**(-1) * a_0 / unit.m**2.5).to(unit.Hz))
```

```python
# estimation of the dB wavelength
scan_freq_max = 600 * unit.MHz
scan_mass_600MHz = const.h * scan_freq_max / const.c**2
scan_mass_4MHz = const.h * 4 * unit.MHz / const.c**2
scan_mass_1kHz = const.h * 1 * unit.kHz / const.c**2
v_esc = (2 * np.abs(earth_pot_unit_mass)) ** 0.5

lambda_dB_600MHz = (const.h / (scan_mass_600MHz * v_esc)).si
lambda_dB_4MHz = (const.h / (scan_mass_4MHz * v_esc)).si
lambda_dB_1kHz = (const.h / (scan_mass_1kHz * v_esc)).si

print("Escape velocity at Earth surface:", v_esc.to(unit.km / unit.s))
print("ALP mass at 600 MHz Compton freq:", (scan_mass_600MHz * const.c**2).to(unit.eV))
print("de Broglie wavelength at 600 MHz:", lambda_dB_600MHz.to(unit.km))
print("ALP mass at 4 MHz Compton freq:", (scan_mass_4MHz * const.c**2).to(unit.eV))
print("de Broglie wavelength at 4 MHz:", lambda_dB_4MHz.to(unit.km), "=", lambda_dB_4MHz.to(unit.earthRad))
# 1 kHz
print("ALP mass at 1 kHz Compton freq:", (scan_mass_1kHz * const.c**2).to(unit.eV))
print(f"de Broglie wavelength at 1 kHz: {lambda_dB_1kHz.to(unit.m):e}", "=", lambda_dB_1kHz.to(unit.earthRad))
```
